```python
# Movie ratings matrix (5 viewers, 3 movies)
movie_ratings = np.array([
    [5, 1, 4],
    [4, 4, 2],
    [4, 3, 5],
    [1, 1, 5],
    [3, 2, 1]
])

# Average rating per movie
average_rating = np.mean(movie_ratings, axis=0)
viewer_preference = np.max(movie_ratings, axis=1)
print("Average Rating:\n", average_rating)
print("Viewer preference:\n", viewer_preference)
```

```python
# Define two matrices
matrix_B = np.array([[2, 3], [4, 5]])
matrix_C = np.array([[1, 1], [1, 1]])

# Adding and subtracting matrices
sum_matrix = np.add(matrix_B, matrix_C)
diff_matrix = np.subtract(matrix_B, matrix_C)

print("Sum of Matrices:\n", sum_matrix)
print("Difference of Matrices:\n", diff_matrix)
```

```python
# Multiplication of two matrices
prod_matrix = np.dot(matrix_B, matrix_C)
print("Product of Matrices:\n", prod_matrix)
```

```python
# Exercise - Matrix Operations
A = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])

# 1. Determinant
# Determinant tells us if the matrix is healthy. No redundant columns, the same, duplicate, multiplication of another column
det_A = np.linalg.det(A)
print(f"Determinant: {det_A:.2f}")

# 2. Inverse
# The inverse of the matrix. If we multiply the matrix by the inverse we get the identity
A_inv = np.linalg.inv(A)
print("\nInverse:")
print(A_inv.round(4))

# 3. Verify
product = A @ A_inv
print(f"\nIs identity? {np.allclose(product, np.eye(3))}")
```

```python
# Exercise - Time Series Manipulation
np.random.seed(99)

dates = np.arange('2023-01-01', '2024-01-01', dtype='datetime64[D]')
prices = 50 + np.cumsum(np.random.normal(0, 1.5, len(dates)))

quarters = [
    ("Q1: Jan-Mar", '2023-01-01', '2023-04-01'),
    ("Q2: Apr-Jun", '2023-04-01', '2023-07-01'),
    ("Q3: Jul-Sep", '2023-07-01', '2023-10-01'),
    ("Q4: Oct-Dec", '2023-10-01', '2024-01-01'),
]

for name, start, end in quarters:
    mask = ((dates >= np.datetime64(start)) & (dates <= np.datetime64(end)) )  # boolean mask for this quarter
    q_data = prices[mask]
    print(f"{name}: Mean={q_data.mean():.2f}, Min={q_data.min():.2f}, Max={q_data.max():.2f}")
```

1. Data Creation Exercise:

Objective: Create a simulated time series dataset.

Instructions:
* Generate a date range from January 1, 2022, to December 31, 2022, with a daily frequency.
* Simulate daily temperature data for this period. Assume temperature varies between 10°C and 35°C.
* Simulate daily humidity data for the same period. Assume humidity varies between 30% and 90%.
* Structure the data into a DataFrame with columns: Date, Temperature, and Humidity.

```python
dates_2022 = np.arange('2022-01-01', '2023-01-01', dtype='datetime64[D]')

temps = np.random.uniform(10,35, size=len(dates))
humidity = np.random.uniform(30, 90, size=len(dates))

dtype = [('Date', 'datetime64[D]'), ('Temperature', 'f4'), ('Humidity', 'f4')]
weather_data = np.zeros(len(dates), dtype=dtype)

weather_data['Date'] = dates_2022
weather_data['Temperature'] = temps
weather_data['Humidity'] = humidity

print(weather_data[:5])
```

```python
temp_mean = np.mean(weather_data['Temperature'])
temp_median = np.median(weather_data['Temperature'])
temp_std = np.std(weather_data['Temperature'])

humidity_mean = np.mean(weather_data['Humidity'])
humidity_median = np.median(weather_data['Humidity'])
humidity_std = np.std(weather_data['Humidity'])

print(f"Temp Mean: {temp_mean:.2f}, Median: {temp_median:.2f}, Std: {temp_std:.2f}")
print(f"Humidity Mean: {humidity_mean: .2f}, Median: {humidity_median: .2f}, std: {humidity_std:.2f}")

max_temp_idx = np.argmax(weather_data['Temperature'])
max_temp_date = weather_data['Date'][max_temp_idx]

max_hum_idx = np.argmax(weather_data['Humidity'])
max_hum_date = weather_data['Date'][max_hum_idx]



```

3. Window Function Exercise:

Objective: Apply window functions to the time series data.

Instructions:
* Use a rolling window of 7 days to calculate the moving average for both temperature and humidity.
* [bonus] Plot the original data and the 7-day moving average on the same graph for comparison.

```python
# Convert the numpy data into a dataframe for pandas
df = pd.DataFrame(weather_data)
# set the date as the index, for the time series power
df.set_index('Date', inplace=True)

df['Temp_7Day_Avg'] = df['Temperature'].rolling(window=7).mean()
df['Humidity_7Day_Avg'] = df['Temperature'].rolling(window=7).mean()

print(df[['Temperature', 'Temp_7Day_Avg']].head(10))
```
