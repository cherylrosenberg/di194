```python
# 1. Unique User-IDs in books_genre
unique_users_rated = books_genre['User-ID'].nunique()
print(f"Unique Users in ratings: {unique_users_rated}")

# 2. Check for missing users 
missing = books_genre[~books_genre['User-ID'].isin(users['User-ID'])]
print(f"User-IDs in ratings NOT in users table: {missing['User-ID'].nunique()}")

# 3. Unique themes count
print(f"Unique Themes: {books_genre['Theme'].nunique()}")
# print(books_genre['Theme'].value_counts().head(25)) # Top 10 genres

# 3. Unique category count
print(f"Unique Categories: {books_genre['category'].nunique()}")
print(books_genre['category'].value_counts().head(10)) # Top 10 genres
```
